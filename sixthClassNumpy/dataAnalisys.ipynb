```python
import numpy as np
```

```python
# Vendas diárias de uma loja (em R$) ao longo de 10 dias
vendas = np.array([1200, 1350, 980, 1500, 1100, 
                    1650, 1400, 1250, 1800, 1300])

print(vendas)
print(f"Tipo: {vendas.dtype}, Formato: {vendas.shape}")
```

```python
print(f"Total vendido: R$ {vendas.sum()}")
print(f"Média diária: R$ {vendas.mean():.2f}")
print(f"Maior venda: R$ {vendas.max()} (dia {vendas.argmax() + 1})")
print(f"Menor venda: R$ {vendas.min()} (dia {vendas.argmin() + 1})")
print(f"Desvio padrão: R$ {vendas.std():.2f}")
```

```python
soma = 0
maxValue = 0
minValue = vendas[0]
for i in vendas:
    print(i)
    soma += i
    if i > maxValue:
        maxValue = i
    if i < minValue:
        minValue = i
mean = soma/len(vendas)
print(f"soma: {soma}")
print(f"media {mean}")
print(f"Maior {maxValue}")
print(f"Menor {minValue}")


```

```python
# Linhas = produtos, Colunas = dias
vendas_produtos = np.array([
    [120, 135, 98, 150, 110],   # Produto A
    [200, 210, 195, 220, 205],  # Produto B
    [80,  90,  75,  85,  95]    # Produto C
])

print(vendas_produtos.shape)  # (3, 5) -> 3 produtos, 5 dias
```

```python
# axis=0 -> desce as colunas (soma por dia, todos produtos)
total_por_dia = vendas_produtos.sum(axis=0)
print("Total por dia:", total_por_dia)

# axis=1 -> percorre as linhas (soma por produto, todos os dias)
total_por_produto = vendas_produtos.sum(axis=1)
print("Total por produto:", total_por_produto)
```

```python
# Quais dias o produto A vendeu mais de 100 unidades?
produto_a = vendas_produtos[0]
dias_bons = produto_a[produto_a > 100]
print("Vendas acima de 100:", dias_bons)

# Quantos dias isso aconteceu?
print("Quantidade de dias:", (produto_a > 100).sum())
```

```python
# Aplicar um aumento de 10% em todas as vendas
vendas_com_reajuste = vendas_produtos * 1.10
print(vendas_com_reajuste)

# Comparar dois produtos, elemento por elemento
diferenca = vendas_produtos[1] - vendas_produtos[0]  # B menos A
print("Diferença B-A por dia:", diferenca)
```

```python
# Padronização: (valor - média) / desvio padrão
vendas_normalizadas = (vendas - vendas.mean()) / vendas.std()
print(vendas_normalizadas)
```

```python
# Dataset: notas de alunos em 4 provas
notas = np.array([
    [7.5, 8.0, 6.5, 9.0],
    [5.5, 6.0, 7.0, 5.0],
    [9.0, 8.5, 9.5, 8.0],
    [4.0, 5.5, 6.0, 5.5]
])

media_por_aluno = notas.mean(axis=1)
print("Média por aluno:", media_por_aluno)

aprovados = media_por_aluno >= 6.0
print("Aprovados (True/False):", aprovados)
print("Quantidade de aprovados:", aprovados.sum())

melhor_aluno = media_por_aluno.argmax()
print(f"Melhor aluno: índice {melhor_aluno}, média {media_por_aluno[melhor_aluno]:.2f}")
```
